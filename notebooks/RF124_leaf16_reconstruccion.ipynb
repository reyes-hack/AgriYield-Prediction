```python
# CELDA 3. Diagnóstico de disponibilidad de variables satelitales

import pandas as pd
import numpy as np

for nombre in ["BASICO", "PRO"]:
    df = datos_oficiales[nombre].copy()

    df["fecha_captura"] = pd.to_datetime(
        df["fecha_captura"],
        dayfirst=True,
        errors="coerce"
    )

    columnas_id = ["ID_POLIGONO", "fecha_captura", "sensor"]
    columnas_variables = [
        c for c in df.columns
        if c not in columnas_id
        and c != "porcentaje_nubosidad"
        and pd.api.types.is_numeric_dtype(df[c])
    ]

    print(f"\nDiagnóstico: {nombre}")
    print(f"Registros: {len(df):,}")
    print(f"Variables numéricas analizadas: {len(columnas_variables)}")

    # Porcentaje de valores faltantes por variable
    faltantes = (
        df[columnas_variables]
        .isna()
        .mean()
        .mul(100)
        .sort_values(ascending=False)
    )

    print("\nPorcentaje de valores faltantes por variable:")
    display(faltantes.to_frame("faltantes_pct").round(2))

    # Cobertura por sensor
    print("\nDisponibilidad por sensor:")
    resumen_sensor = (
        df.groupby("sensor")[columnas_variables]
        .agg(lambda s: s.notna().mean() * 100)
        .T
    )
    display(resumen_sensor.round(2))

    # Cobertura por año
    df["ANIO"] = df["fecha_captura"].dt.year

    print("\nDisponibilidad por año:")
    resumen_anio = (
        df.groupby("ANIO")[columnas_variables]
        .agg(lambda s: s.notna().mean() * 100)
        .T
    )
    display(resumen_anio.round(2))
```

```python
# CELDA 4. Investigación de patrones de faltantes por sensor,
# periodo y parcela. No modifica los archivos oficiales.

import pandas as pd
import numpy as np

for nombre in ["BASICO", "PRO"]:
    df = datos_oficiales[nombre].copy()

    df["fecha_captura"] = pd.to_datetime(
        df["fecha_captura"],
        dayfirst=True,
        errors="coerce"
    )

    df["ANIO"] = df["fecha_captura"].dt.year
    df["MES"] = df["fecha_captura"].dt.month

    columnas_id = {
        "ID_POLIGONO", "fecha_captura", "sensor", "ANIO", "MES"
    }
    variables = [
        c for c in df.columns
        if c not in columnas_id
        and pd.api.types.is_numeric_dtype(df[c])
    ]

    print(f"\n{nombre}: REGISTROS POR SENSOR Y AÑO")
    display(
        df.groupby(["sensor", "ANIO"])
        .size()
        .rename("registros")
        .reset_index()
    )

    print(f"\n{nombre}: DISPONIBILIDAD MENSUAL")
    mensual = (
        df.groupby(["ANIO", "MES"])[variables]
        .apply(lambda bloque: bloque.notna().mean().mean() * 100)
        .rename("disponibilidad_media_pct")
        .reset_index()
    )
    display(mensual.round(2))

    print(f"\n{nombre}: COBERTURA POR PARCELA")
    cobertura = (
        df.groupby("ID_POLIGONO")[variables]
        .agg(lambda columna: columna.notna().mean() * 100)
        .mean(axis=1)
        .rename("disponibilidad_media_pct")
        .reset_index()
        .sort_values("disponibilidad_media_pct")
    )

    display(cobertura.head(10).round(2))
    display(cobertura.tail(10).round(2))

    print(f"\n{nombre}: OBSERVACIONES CON TODAS LAS VARIABLES VACÍAS")
    filas_vacias = df[variables].isna().all(axis=1)
    print(f"{filas_vacias.sum():,} de {len(df):,} registros")
    print(
        f"Porcentaje: {filas_vacias.mean() * 100:.2f}%"
    )
```

```python
# CELDA 5. Coincidencia de faltantes y unicidad temporal

for nombre in ["BASICO", "PRO"]:
    df = datos_oficiales[nombre].copy()

    columnas_id = ["ID_POLIGONO", "fecha_captura", "sensor"]
    variables = [
        c for c in df.columns
        if c not in columnas_id + ["porcentaje_nubosidad"]
        and pd.api.types.is_numeric_dtype(df[c])
    ]

    faltantes = df[variables].isna()

    print(f"\n{nombre}")
    print("Filas:", len(df))
    print("Variables numéricas:", len(variables))

    # Número de variables vacías por fila
    conteo_vacias = faltantes.sum(axis=1)
    print("\nDistribución del número de variables vacías por fila:")
    display(
        conteo_vacias.value_counts()
        .sort_index()
        .rename_axis("variables_vacias")
        .to_frame("filas")
    )

    # Comprobar si los patrones de faltantes son idénticos
    patrones = faltantes.drop_duplicates()
    print("Patrones distintos de faltantes:", len(patrones))

    # Unicidad de las observaciones por parcela, fecha y sensor
    duplicados_clave = df.duplicated(
        subset=columnas_id,
        keep=False
    )

    print(
        "Filas implicadas en duplicados parcela-fecha-sensor:",
        int(duplicados_clave.sum())
    )

    # Comparación de cobertura por sensor
    if "sensor" in df.columns:
        print("\nCobertura de variables por sensor:")
        cobertura_sensor = (
            df.groupby("sensor")[variables]
            .apply(lambda grupo: grupo.notna().mean().mean() * 100)
        )
        display(cobertura_sensor.round(2).to_frame(
            "disponibilidad_media_pct"
        ))
```

```python
# CELDA 6. Auditoría específica por sensor y filas atípicas

basico = datos_oficiales["BASICO"].copy()

columnas_id = [
    "ID_POLIGONO",
    "fecha_captura",
    "sensor"
]

variables = [
    c for c in basico.columns
    if c not in columnas_id
    and pd.api.types.is_numeric_dtype(basico[c])
]

for sensor, grupo in basico.groupby("sensor"):
    print(f"\nSENSOR: {sensor}")
    print(f"Registros: {len(grupo):,}")

    auditoria = pd.DataFrame({
        "registros_con_dato": grupo[variables].notna().sum(),
        "disponibilidad_pct": grupo[variables].notna().mean() * 100,
        "valores_distintos": grupo[variables].nunique(dropna=True)
    }).sort_values(
        ["disponibilidad_pct", "registros_con_dato"],
        ascending=False
    )

    display(auditoria.round(2))

# Inspeccionar las cuatro filas con exactamente ocho variables vacías
conteo_vacias = basico[variables].isna().sum(axis=1)
atipicas = basico.loc[conteo_vacias == 8].copy()

print("\nFilas con exactamente ocho variables vacías:")
display(atipicas)

# Inspeccionar las columnas que sí contienen datos en esas filas
print("\nVariables disponibles en las filas atípicas:")
display(
    atipicas[variables]
    .notna()
    .sum()
    .loc[lambda s: s > 0]
    .rename("registros_con_dato")
    .to_frame()
)
```

```python
# CELDA 7. Cobertura anual por sensor y parcela

basico = datos_oficiales["BASICO"].copy()

basico["fecha_captura"] = pd.to_datetime(
    basico["fecha_captura"],
    dayfirst=True,
    errors="coerce"
)
basico["ANIO"] = basico["fecha_captura"].dt.year

variables_por_sensor = {
    "Landsat": [
        "vi6t_promedio",
        "vi6t_std",
        "vi6t_max",
        "vi6t_min",
    ],
    "Sentinel-2": [
        "ndvi_promedio",
        "evi_promedio",
        "lai_promedio",
        "fapar_promedio",
        "ndwi_promedio",
        "nddi_promedio",
    ],
}

for sensor, variables in variables_por_sensor.items():
    sub = basico.loc[basico["sensor"] == sensor].copy()

    # Usamos solo variables que realmente existen en la fuente.
    variables = [v for v in variables if v in sub.columns]

    # Una observación es válida si al menos una variable seleccionada
    # tiene un valor numérico.
    sub["observacion_valida"] = sub[variables].notna().any(axis=1)

    print(f"\n{sensor}: observaciones válidas por año")
    display(
        sub.groupby("ANIO")
        .agg(
            registros=("ID_POLIGONO", "size"),
            parcelas=("ID_POLIGONO", "nunique"),
            observaciones_validas=("observacion_valida", "sum"),
        )
        .assign(
            disponibilidad_pct=lambda x:
                100 * x["observaciones_validas"] / x["registros"]
        )
        .round(2)
    )

    print(f"\n{sensor}: cobertura por parcela y año")
    cobertura = (
        sub.groupby(["ID_POLIGONO", "ANIO"])
        .agg(
            registros=("ID_POLIGONO", "size"),
            observaciones_validas=("observacion_valida", "sum"),
        )
        .reset_index()
    )

    cobertura["disponibilidad_pct"] = (
        100 * cobertura["observaciones_validas"]
        / cobertura["registros"]
    )

    display(
        cobertura.sort_values(
            ["ANIO", "disponibilidad_pct"]
        ).head(15).round(2)
    )
```

```python
# CELDA 8. Reconstrucción de las 51 variables satelitales Planet
# Fuente: archivos oficiales del reto.
# No utiliza tablas intermedias generadas en experimentos anteriores.

import pandas as pd
import numpy as np

from pathlib import Path

# 1. Definir las ventanas originales documentadas
VENTANAS = {
    "inicial": ("2025-04-01", "2025-05-31"),
    "intermedia": ("2025-06-01", "2025-07-31"),
    "tardia": ("2025-08-01", "2025-10-31"),
}

INDICES = ["ndvi", "evi", "lai", "msavi"]
ESTADISTICAS = ["promedio", "std", "max", "min"]

# 2. Leer nuevamente la fuente oficial PRO desde disco
ruta_pro = FUENTES_OFICIALES["PRO"]

pro_original = pd.read_csv(
    ruta_pro,
    dtype={"fecha_captura": "string"},
    low_memory=False
)

pro_original.columns = pro_original.columns.str.strip()

pro_original["ID_POLIGONO"] = (
    pro_original["ID_POLIGONO"].astype(str).str.strip()
)

pro_original["fecha_captura"] = pd.to_datetime(
    pro_original["fecha_captura"].str.strip(),
    format="%d/%m/%Y",
    errors="raise"
)

# Validaciones de la fuente
assert pro_original["ID_POLIGONO"].notna().all()
assert pro_original["fecha_captura"].notna().all()
assert not pro_original.duplicated(
    ["ID_POLIGONO", "fecha_captura", "sensor"]
).any()

pro_2025 = pro_original.loc[
    (pro_original["fecha_captura"].dt.year == 2025)
    & (pro_original["sensor"].eq("Planet"))
].copy()

assert pro_2025["ID_POLIGONO"].nunique() == 197

# 3. Verificar las 16 variables espectrales esperadas
VARIABLES_ESPECTRALES = [
    f"{indice}_{estadistica}"
    for indice in INDICES
    for estadistica in ESTADISTICAS
]

faltantes_columnas = sorted(
    set(VARIABLES_ESPECTRALES) - set(pro_2025.columns)
)

if faltantes_columnas:
    raise ValueError(
        f"Faltan variables espectrales oficiales: {faltantes_columnas}"
    )

# 4. Reconstruir cada ventana desde las observaciones originales
agregados = []

for nombre_ventana, (inicio, fin) in VENTANAS.items():
    inicio = pd.Timestamp(inicio)
    fin = pd.Timestamp(fin)

    bloque = pro_2025.loc[
        pro_2025["fecha_captura"].between(inicio, fin)
    ].copy()

    print(
        f"{nombre_ventana}: {len(bloque):,} observaciones, "
        f"{bloque['ID_POLIGONO'].nunique()} parcelas"
    )

    # Promedio temporal de las 16 estadísticas originales
    espectrales = (
        bloque.groupby("ID_POLIGONO")[VARIABLES_ESPECTRALES]
        .mean()
        .add_suffix(f"_{nombre_ventana}")
    )

    # Conteo de observaciones por parcela en la ventana
    cobertura = (
        bloque.groupby("ID_POLIGONO")
        .size()
        .rename(f"observaciones_{nombre_ventana}")
    )

    agregado = espectrales.join(cobertura, how="outer")
    agregados.append(agregado)

# 5. Unir las tres ventanas y conservar todos los IDs oficiales
planet_features_reconstruido = (
    pd.concat(agregados, axis=1)
    .reset_index()
)

planet_features_reconstruido["ID_POLIGONO"] = (
    planet_features_reconstruido["ID_POLIGONO"].astype(str).str.strip()
)

ids_oficiales = set(
    etiquetas["ID_POLIGONO"].astype(str).str.strip()
)

ids_reconstruidos = set(
    planet_features_reconstruido["ID_POLIGONO"]
)

if ids_reconstruidos != ids_oficiales:
    raise ValueError(
        "Los identificadores de Planet no coinciden con los oficiales. "
        f"Faltan: {len(ids_oficiales - ids_reconstruidos)}; "
        f"sobran: {len(ids_reconstruidos - ids_oficiales)}."
    )

if planet_features_reconstruido["ID_POLIGONO"].duplicated().any():
    raise ValueError("La tabla Planet reconstruida tiene IDs duplicados.")

# 6. Verificar la estructura esperada: 48 estadísticas + 3 conteos
features_planet = [
    c for c in planet_features_reconstruido.columns
    if c != "ID_POLIGONO"
]

assert len(features_planet) == 51, (
    f"Se esperaban 51 variables satelitales; "
    f"se obtuvieron {len(features_planet)}."
)

print("\nReconstrucción satelital terminada")
print("Dimensiones:", planet_features_reconstruido.shape)
print("Parcelas:", planet_features_reconstruido["ID_POLIGONO"].nunique())
print("Variables satelitales:", len(features_planet))

print("\nValores faltantes por variable:")
display(
    planet_features_reconstruido[features_planet]
    .isna()
    .sum()
    .loc[lambda s: s > 0]
    .to_frame("faltantes")
)

display(planet_features_reconstruido.head())
```

```python
# CELDA 9. Integración de variables satelitales con etiquetas oficiales
# No utiliza matrices ni tablas intermedias de experimentos anteriores.

import pandas as pd
import numpy as np

etiquetas_oficiales = pd.read_csv(
    FUENTES_OFICIALES["ID_RENDIMIENTO"]
)

etiquetas_oficiales.columns = etiquetas_oficiales.columns.str.strip()
etiquetas_oficiales["ID_POLIGONO"] = (
    etiquetas_oficiales["ID_POLIGONO"].astype(str).str.strip()
)

# Validar la fuente oficial
assert len(etiquetas_oficiales) == 197
assert etiquetas_oficiales["ID_POLIGONO"].is_unique
assert etiquetas_oficiales["AREA_HA"].notna().all()
assert etiquetas_oficiales["CONJUNTO"].notna().all()

# Unir mediante el identificador, nunca por posición
matriz_base = etiquetas_oficiales[
    ["ID_POLIGONO", "AREA_HA", "RENDIMIENTO_T_HA", "CONJUNTO"]
].merge(
    planet_features_reconstruido,
    on="ID_POLIGONO",
    how="left",
    validate="one_to_one",
    indicator=True,
)

print("Coincidencia de IDs:")
print(matriz_base["_merge"].value_counts().to_dict())

assert matriz_base["_merge"].eq("both").all()
assert matriz_base["ID_POLIGONO"].is_unique
assert len(matriz_base) == 197

matriz_base = matriz_base.drop(columns="_merge")

columnas_satelitales = [
    c for c in planet_features_reconstruido.columns
    if c != "ID_POLIGONO"
]

print("\nDimensiones de la matriz base:", matriz_base.shape)
print("Variables satelitales:", len(columnas_satelitales))
print("Parcelas de entrenamiento:",
      matriz_base["RENDIMIENTO_T_HA"].notna().sum())
print("Parcelas de predicción:",
      matriz_base["RENDIMIENTO_T_HA"].isna().sum())
print("Faltantes en variables satelitales:",
      int(matriz_base[columnas_satelitales].isna().sum().sum()))
print("Faltantes en AREA_HA:",
      int(matriz_base["AREA_HA"].isna().sum()))

display(matriz_base.head())
```

```python
# CELDA 10. Inventario de posibles fuentes climáticas originales
# Solo inspección: no lee ni utiliza archivos intermedios.

from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent
RAW = ROOT / "data" / "raw"

EXTENSIONES = {
    ".tif", ".tiff", ".nc", ".grib", ".grb",
    ".csv", ".parquet", ".shp", ".gpkg",
    ".zip", ".hdf", ".h5", ".asc"
}

archivos = sorted(
    p for p in RAW.rglob("*")
    if p.is_file() and p.suffix.lower() in EXTENSIONES
)

inventario = []

for p in archivos:
    relativo = p.relative_to(ROOT).as_posix()
    nombre = p.name.lower()
    ruta_lower = relativo.lower()

    if any(k in nombre or k in ruta_lower for k in [
        "chirps", "precip", "rain", "era5", "chirts",
        "tmax", "tmin", "temperatura", "temperature",
        "climate", "clima"
    ]):
        categoria = "Clima"
    elif any(k in nombre or k in ruta_lower for k in [
        "agrocebada", "rendimiento", "area"
    ]):
        categoria = "Reto oficial / etiquetas"
    else:
        categoria = "Otra fuente raw"

    inventario.append({
        "archivo": relativo,
        "extension": p.suffix.lower(),
        "tamano_MB": round(p.stat().st_size / (1024 ** 2), 3),
        "categoria": categoria,
    })

inventario_raw = pd.DataFrame(inventario)

print("Raíz inspeccionada:", RAW)
print("Archivos tabulares/geoespaciales identificados:", len(archivos))

if inventario_raw.empty:
    print("No se encontraron archivos candidatos en data/raw.")
else:
    display(
        inventario_raw.sort_values(
            ["categoria", "archivo"]
        ).reset_index(drop=True)
    )

print("\nFuentes climáticas identificadas:")
if inventario_raw.empty:
    print("Ninguna.")
else:
    display(
        inventario_raw.loc[
            inventario_raw["categoria"].eq("Clima")
        ].reset_index(drop=True)
    )
```

```python

# CELDA 11. Inspección de fuentes climáticas originales

from pathlib import Path
import re
import pandas as pd

ROOT = Path.cwd().parent
RAW = ROOT / "data" / "raw"

carpetas_clima = [
    RAW / "climate" / "precipitation",
    RAW / "climate" / "temperature",
]

registros = []

for carpeta in carpetas_clima:
    if not carpeta.exists():
        print(f"ADVERTENCIA: no existe {carpeta.relative_to(ROOT)}")
        continue

    for archivo in sorted(carpeta.rglob("*")):
        if not archivo.is_file():
            continue

        nombre = archivo.name
        ruta_relativa = archivo.relative_to(ROOT).as_posix()

        # Extraer años y meses cuando estén presentes en el nombre
        anios = re.findall(r"20\d{2}", nombre)
        meses = re.findall(
            r"(?<!\d)(0?[1-9]|1[0-2])(?!\d)", nombre
        )

        registros.append({
            "archivo": ruta_relativa,
            "nombre": nombre,
            "carpeta": archivo.parent.relative_to(ROOT).as_posix(),
            "extension": archivo.suffix.lower(),
            "tamano_kb": round(archivo.stat().st_size / 1024, 2),
            "anios_en_nombre": ", ".join(dict.fromkeys(anios)),
            "posibles_meses_en_nombre": ", ".join(dict.fromkeys(meses)),
        })

inventario_climatico = pd.DataFrame(registros)

print("Total de archivos climáticos:", len(inventario_climatico))

if not inventario_climatico.empty:
    display(
        inventario_climatico
        .groupby(["carpeta", "extension"], dropna=False)
        .agg(
            cantidad=("archivo", "size"),
            tamano_total_kb=("tamano_kb", "sum"),
        )
        .reset_index()
    )

    print("\nPrimeros 30 archivos:")
    display(inventario_climatico.head(30))

    print("\nÚltimos 20 archivos:")
    display(inventario_climatico.tail(20))
```

```python

# CELDA 12. Auditoría de metadatos de los rásteres climáticos

from pathlib import Path
import pandas as pd
import rasterio
import importlib.util

ROOT = Path.cwd().parent
RAW = ROOT / "data" / "raw"

if importlib.util.find_spec("rasterio") is None:
    raise ImportError(
        "Falta rasterio. Instálalo en el entorno del notebook con: "
        "%pip install rasterio"
    )

carpetas = {
    "precipitacion": RAW / "climate" / "precipitation",
    "temperatura_maxima": RAW / "climate" / "temperature" / "Tmax",
    "temperatura_minima": RAW / "climate" / "temperature" / "Tmin",
}

metadatos = []

for variable, carpeta in carpetas.items():
    for archivo in sorted(carpeta.rglob("*.tif")):
        with rasterio.open(archivo) as src:
            banda = src.read(1, masked=True)
            valores = banda.compressed()

            metadatos.append({
                "variable": variable,
                "archivo": archivo.relative_to(ROOT).as_posix(),
                "ancho_px": src.width,
                "alto_px": src.height,
                "bandas": src.count,
                "crs": str(src.crs),
                "resolucion_x": src.res[0],
                "resolucion_y": src.res[1],
                "nodata": src.nodata,
                "pixeles_validos": int(valores.size),
                "minimo": float(valores.min()) if valores.size else None,
                "maximo": float(valores.max()) if valores.size else None,
                "media": float(valores.mean()) if valores.size else None,
            })

auditoria_raster = pd.DataFrame(metadatos)

print("Número de rásteres inspeccionados:", len(auditoria_raster))
print("\nResumen por variable:")

display(
    auditoria_raster.groupby("variable").agg(
        archivos=("archivo", "count"),
        sistemas_coordenadas=("crs", "nunique"),
        resoluciones=("resolucion_x", "nunique"),
        tamanos=("ancho_px", "nunique"),
        valores_minimos=("minimo", "min"),
        valores_maximos=("maximo", "max"),
    )
)

print("\nMuestra de archivos de cada variable:")

display(
    auditoria_raster.groupby("variable", group_keys=False)
    .head(2)
    .reset_index(drop=True)
)
```

```python
# ============================================================
# CELDA 13 — VALIDACIÓN DE RÁSTERES CLIMÁTICOS ORIGINALES
# Proyecto: Reto AgroCebada FIRA 2026
#
# Independencia:
# - Solo lee archivos originales de data/raw.
# - No consulta otros notebooks.
# - No carga matrices ni resultados intermedios.
# - No modifica ningún archivo.
# ============================================================

from pathlib import Path
import re
import pandas as pd

# ------------------------------------------------------------
# 1. LOCALIZAR LAS FUENTES ORIGINALES
# ------------------------------------------------------------

if "ROOT" not in globals():
    raise NameError(
        "No existe ROOT. Ejecuta primero la celda 1."
    )

RAW_DIR = Path(ROOT) / "data" / "raw"

if not RAW_DIR.is_dir():
    raise FileNotFoundError(
        f"No se encuentra el directorio de fuentes originales: {RAW_DIR}"
    )

# ------------------------------------------------------------
# 2. INVENTARIAR ARCHIVOS RÁSTER ORIGINALES
# ------------------------------------------------------------

archivos_tif = sorted(
    p for p in RAW_DIR.rglob("*")
    if p.is_file()
    and p.suffix.lower() in {".tif", ".tiff"}
)

if not archivos_tif:
    raise FileNotFoundError(
        f"No se encontraron GeoTIFF originales en {RAW_DIR}"
    )

print("Directorio de fuentes originales:", RAW_DIR)
print("GeoTIFF encontrados:", len(archivos_tif))

# ------------------------------------------------------------
# 3. IDENTIFICAR VARIABLES Y PERIODOS POR NOMBRE
# ------------------------------------------------------------

patron_fecha = re.compile(
    r"(?P<year>20\d{2})[_-](?P<month>0[1-9]|1[0-2])"
)

registros = []

for ruta in archivos_tif:
    nombre = ruta.stem.upper()

    if "PREC" in nombre or "PRECIP" in nombre:
        variable = "precipitacion"
    elif "TMAX" in nombre or "T_MAX" in nombre:
        variable = "temperatura_maxima"
    elif "TMIN" in nombre or "T_MIN" in nombre:
        variable = "temperatura_minima"
    else:
        variable = "sin_clasificar"

    coincidencia = patron_fecha.search(ruta.stem)

    if coincidencia:
        anio = int(coincidencia.group("year"))
        mes = int(coincidencia.group("month"))
    else:
        anio = None
        mes = None

    registros.append({
        "archivo": ruta.name,
        "ruta_relativa": str(ruta.relative_to(RAW_DIR)),
        "variable": variable,
        "anio": anio,
        "mes": mes,
        "bytes": ruta.stat().st_size,
    })

inventario_climatico = pd.DataFrame(registros)

# ------------------------------------------------------------
# 4. MOSTRAR COBERTURA TEMPORAL
# ------------------------------------------------------------

print("\n=== ARCHIVOS POR VARIABLE ===")
display(
    inventario_climatico
    .groupby("variable")
    .size()
    .rename("numero_archivos")
    .to_frame()
)

print("\n=== ARCHIVOS POR VARIABLE Y AÑO ===")
display(
    inventario_climatico
    .pivot_table(
        index="anio",
        columns="variable",
        values="archivo",
        aggfunc="count",
        fill_value=0
    )
    .sort_index()
)

print("\n=== PERIODOS IDENTIFICADOS ===")
display(
    inventario_climatico[
        ["archivo", "variable", "anio", "mes", "bytes"]
    ].sort_values(
        ["variable", "anio", "mes", "archivo"],
        na_position="last"
    ).reset_index(drop=True)
)

# ------------------------------------------------------------
# 5. AUDITORÍA DE INTEGRIDAD DEL INVENTARIO
# ------------------------------------------------------------

print("\n=== AUDITORÍA ===")
print("Archivos inventariados:", len(inventario_climatico))
print(
    "Archivos sin año o mes identificable:",
    int(inventario_climatico[["anio", "mes"]].isna().any(axis=1).sum())
)
print(
    "Archivos sin clasificar:",
    int((inventario_climatico["variable"] == "sin_clasificar").sum())
)
print(
    "Archivos vacíos:",
    int((inventario_climatico["bytes"] == 0).sum())
)

print(
    "\nNota: el nombre del archivo permite identificar candidatos, "
    "pero no confirma por sí solo las unidades, acumulaciones "
    "ni el periodo agronómico de cada variable."
)
```

```python
# CELDA 16 — COBERTURA MENSUAL POR SENSOR
# Fuentes: CSV oficiales BASICO y PRO exclusivamente.

import pandas as pd

for nombre in ["BASICO", "PRO"]:
    df = datos_oficiales[nombre].copy()

    # Las fechas oficiales pueden venir como DD/MM/YYYY.
    df["fecha_captura"] = pd.to_datetime(
        df["fecha_captura"],
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )

    print(f"\n{'=' * 65}")
    print(f"FUENTE: {nombre}")

    # Número de registros por año y mes.
    cobertura = (
        df.dropna(subset=["fecha_captura"])
        .assign(
            anio=lambda x: x["fecha_captura"].dt.year,
            mes=lambda x: x["fecha_captura"].dt.month
        )
        .groupby(["sensor", "anio", "mes"])
        .agg(
            registros=("ID_POLIGONO", "size"),
            parcelas=("ID_POLIGONO", "nunique"),
            fechas=("fecha_captura", "nunique")
        )
        .reset_index()
        .sort_values(["sensor", "anio", "mes"])
    )

    display(cobertura)

    print(
        "\nRegistros con fecha inválida:",
        int(df["fecha_captura"].isna().sum())
    )
```

```python
# CELDA 20 — COBERTURA ESPACIAL DE PARCELAS Y RÁSTERES CLIMÁTICOS

from pathlib import Path
import rasterio
import geopandas as gpd

raw_dir = Path(ROOT) / "data" / "raw"

# Usamos exclusivamente el shapefile original y un ráster climático original.
rasteres_clima = sorted(
    p for p in raw_dir.rglob("*")
    if p.is_file()
    and p.suffix.lower() in {".tif", ".tiff"}
    and p.name.lower().startswith(("prec_", "precip_", "tmax_", "tmin_"))
)

if not rasteres_clima:
    raise FileNotFoundError("No se encontraron rásteres climáticos originales.")

raster_ref = rasteres_clima[0]

with rasterio.open(raster_ref) as src:
    raster_crs = src.crs
    raster_bounds = src.bounds
    raster_res = src.res
    raster_shape = (src.height, src.width)
    raster_nodata = src.nodata

print("Ráster de referencia:", raster_ref.name)
print("CRS:", raster_crs)
print("Extensión:", raster_bounds)
print("Resolución:", raster_res)
print("Dimensiones:", raster_shape)
print("NoData:", raster_nodata)

parcelas_clima = parcelas_geo.to_crs(raster_crs)

print("\nCobertura de parcelas:")
print("Parcelas:", len(parcelas_clima))
print("Dentro de la extensión del ráster:",
      int(parcelas_clima.geometry.apply(
          lambda g: (
              g.bounds[0] >= raster_bounds.left
              and g.bounds[1] >= raster_bounds.bottom
              and g.bounds[2] <= raster_bounds.right
              and g.bounds[3] <= raster_bounds.top
          )
      ).sum()))
print("Fuera o parcialmente fuera de la extensión:",
      int((~parcelas_clima.geometry.apply(
          lambda g: (
              g.bounds[0] >= raster_bounds.left
              and g.bounds[1] >= raster_bounds.bottom
              and g.bounds[2] <= raster_bounds.right
              and g.bounds[3] <= raster_bounds.top
          )
      )).sum()))

print("\nExtensión conjunta de parcelas:")
print(parcelas_clima.total_bounds)
```

```python

# CELDA 23 — AUDITORÍA DE PARCELAS, FECHAS Y RENDIMIENTO

from pathlib import Path
import pandas as pd

tabular_dir = Path(ROOT) / "data" / "raw" / "tabular"

basico_path = tabular_dir / "Conjunto_datos_BASICO_AgroCebada2026.csv"
pro_path = tabular_dir / "Conjunto_datos_PRO_AgroCebada.csv"
rend_path = tabular_dir / "ID_area_rendimiento_70_30_Reto_AgroCebada.csv"

basico = pd.read_csv(basico_path)
pro = pd.read_csv(pro_path)
rend = pd.read_csv(rend_path)

# Normalizar IDs solo para la auditoría; no modificar los archivos fuente.
for df in (basico, pro, rend):
    df["ID_POLIGONO"] = df["ID_POLIGONO"].astype("string").str.strip()

for df in (basico, pro):
    df["fecha_captura"] = pd.to_datetime(
        df["fecha_captura"], dayfirst=True, errors="coerce"
    )

ids_basico = set(basico["ID_POLIGONO"].dropna().unique())
ids_pro = set(pro["ID_POLIGONO"].dropna().unique())
ids_rend = set(rend["ID_POLIGONO"].dropna().unique())

ids_objetivo = set(
    rend.loc[rend["RENDIMIENTO_T_HA"].notna(), "ID_POLIGONO"]
)
ids_prediccion = set(
    rend.loc[rend["RENDIMIENTO_T_HA"].isna(), "ID_POLIGONO"]
)

print("=== PARTICIÓN DE RENDIMIENTO ===")
print("Parcelas totales:", rend["ID_POLIGONO"].nunique())
print("Con rendimiento:", len(ids_objetivo))
print("Sin rendimiento:", len(ids_prediccion))
print("\nCONJUNTO:")
print(rend["CONJUNTO"].value_counts(dropna=False).to_string())

print("\n=== COBERTURA DE IDS ===")
for nombre, ids in [
    ("BASIC", ids_basico),
    ("PRO", ids_pro),
]:
    print(f"\n{nombre}")
    print("IDs únicos:", len(ids))
    print("IDs del archivo de rendimiento presentes:", len(ids & ids_rend))
    print("IDs de entrenamiento presentes:", len(ids & ids_objetivo))
    print("IDs de predicción presentes:", len(ids & ids_prediccion))
    print("IDs del archivo de rendimiento sin observaciones:", 
          sorted(ids_rend - ids)[:20])

print("\n=== RESUMEN TEMPORAL Y POR SENSOR ===")
for nombre, df in [("BASIC", basico), ("PRO", pro)]:
    print(f"\n{nombre}")
    print("Fechas inválidas:", int(df["fecha_captura"].isna().sum()))
    print("Fecha mínima:", df["fecha_captura"].min())
    print("Fecha máxima:", df["fecha_captura"].max())
    print("Registros por sensor:")
    print(df["sensor"].value_counts(dropna=False).to_string())

    por_parcela = df.groupby("ID_POLIGONO").agg(
        observaciones=("fecha_captura", "size"),
        fechas_validas=("fecha_captura", "count"),
        fechas_distintas=("fecha_captura", "nunique"),
    )

    print("\nObservaciones por parcela:")
    print(por_parcela["observaciones"].describe().round(2).to_string())

    duplicados = df.duplicated(
        subset=["ID_POLIGONO", "fecha_captura", "sensor"],
        keep=False
    )
    print("\nFilas que comparten ID, fecha y sensor:", int(duplicados.sum()))

print("\n=== CONTROL DEL ARCHIVO DE RENDIMIENTO ===")
print("IDs duplicados:", int(rend["ID_POLIGONO"].duplicated().sum()))
print("Rendimientos negativos:",
      int((rend["RENDIMIENTO_T_HA"].dropna() < 0).sum()))
print("Particiones desconocidas:",
      sorted(set(rend["CONJUNTO"].dropna().astype(str)) -
             {"ENTRENAMIENTO", "PREDICCION"}))

print("\nAuditoría terminada. Los archivos originales no fueron modificados.")
```

```python

# CELDA 24 — DIAGNÓSTICO TEMPORAL Y CALIDAD DE LOS DATOS

import pandas as pd
from pathlib import Path

tabular_dir = Path(ROOT) / "data" / "raw" / "tabular"

basico = pd.read_csv(
    tabular_dir / "Conjunto_datos_BASICO_AgroCebada2026.csv"
)
pro = pd.read_csv(
    tabular_dir / "Conjunto_datos_PRO_AgroCebada.csv"
)

for df in (basico, pro):
    df["fecha_captura"] = pd.to_datetime(
        df["fecha_captura"], dayfirst=True, errors="coerce"
    )
    df["año"] = df["fecha_captura"].dt.year
    df["mes"] = df["fecha_captura"].dt.month

# 1. Cantidad de observaciones por año y sensor
print("=" * 80)
print("1. OBSERVACIONES POR AÑO Y SENSOR")

for nombre, df in [("BASIC", basico), ("PRO", pro)]:
    print(f"\n--- {nombre} ---")
    print(
        df.groupby(["año", "sensor"], dropna=False)
          .size()
          .unstack(fill_value=0)
          .to_string()
    )

# 2. Cobertura mensual por año y sensor
print("\n" + "=" * 80)
print("2. MESES CON OBSERVACIONES POR AÑO Y SENSOR")

for nombre, df in [("BASIC", basico), ("PRO", pro)]:
    print(f"\n--- {nombre} ---")
    cobertura = (
        df.groupby(["año", "sensor"])["mes"]
          .nunique()
          .rename("meses_con_observaciones")
    )
    print(cobertura.to_string())

# 3. Proporción de valores faltantes por sensor
print("\n" + "=" * 80)
print("3. VALORES FALTANTES POR SENSOR")

columnas_numericas_basico = [
    c for c in basico.select_dtypes(include="number").columns
    if c not in {"año", "mes"}
]
columnas_numericas_pro = [
    c for c in pro.select_dtypes(include="number").columns
    if c not in {"año", "mes"}
]

for nombre, df, columnas in [
    ("BASIC", basico, columnas_numericas_basico),
    ("PRO", pro, columnas_numericas_pro),
]:
    print(f"\n--- {nombre} ---")
    faltantes = df.groupby("sensor")[columnas].agg(
        lambda s: s.isna().mean() * 100
    )
    print("Porcentaje de valores faltantes:")
    print(faltantes.round(1).to_string())

# 4. Distribución de registros por parcela y sensor
print("\n" + "=" * 80)
print("4. REGISTROS POR PARCELA Y SENSOR")

for nombre, df in [("BASIC", basico), ("PRO", pro)]:
    print(f"\n--- {nombre} ---")
    conteos = df.groupby(["ID_POLIGONO", "sensor"]).size()
    print(conteos.groupby("sensor").describe().round(2).to_string())

# 5. Número de observaciones completamente vacías en las variables
# espectrales, sin contar identificadores, fecha, sensor ni nubosidad.
print("\n" + "=" * 80)
print("5. FILAS SIN NINGÚN ÍNDICE ESPECTRAL DISPONIBLE")

for nombre, df in [("BASIC", basico), ("PRO", pro)]:
    excluir = {
        "ID_POLIGONO", "fecha_captura", "sensor",
        "año", "mes", "porcentaje_nubosidad"
    }
    indices = [
        c for c in df.select_dtypes(include="number").columns
        if c not in excluir
    ]

    filas_vacias = df[indices].isna().all(axis=1)

    print(
        f"{nombre}: {int(filas_vacias.sum()):,} de {len(df):,} filas "
        f"({filas_vacias.mean() * 100:.2f}%)"
    )

print("\nDiagnóstico terminado; los archivos originales no fueron modificados.")
```

```python

# CELDA 25 — DISPONIBILIDAD TEMPORAL DE ÍNDICES ESPECTRALES

import pandas as pd
from pathlib import Path

tabular_dir = Path(ROOT) / "data" / "raw" / "tabular"

basico = pd.read_csv(
    tabular_dir / "Conjunto_datos_BASICO_AgroCebada2026.csv"
)

basico["fecha_captura"] = pd.to_datetime(
    basico["fecha_captura"], dayfirst=True, errors="coerce"
)
basico["año"] = basico["fecha_captura"].dt.year
basico["mes"] = basico["fecha_captura"].dt.month

# Índices representativos que aparecen en el archivo original.
indices = [
    "ndvi_promedio",
    "evi_promedio",
    "lai_promedio",
    "savi_promedio",
    "vi6t_promedio",
    "ndwi_promedio",
    "ndti_promedio",
]

print("=== DISPONIBILIDAD POR AÑO Y SENSOR ===")

for indice in indices:
    print(f"\n--- {indice} ---")

    disponibilidad = (
        basico.groupby(["año", "sensor"])[indice]
        .agg(
            registros="size",
            validos="count",
        )
        .reset_index()
    )

    disponibilidad["porcentaje_valido"] = (
        disponibilidad["validos"] /
        disponibilidad["registros"] * 100
    )

    print(
        disponibilidad.to_string(
            index=False,
            formatters={
                "porcentaje_valido": lambda x: f"{x:.1f}%"
            }
        )
    )

print("\n=== DISPONIBILIDAD MENSUAL DE NDVI ===")

ndvi_mensual = (
    basico.groupby(["sensor", "año", "mes"])["ndvi_promedio"]
    .agg(registros="size", validos="count")
    .reset_index()
)

ndvi_mensual["porcentaje_valido"] = (
    ndvi_mensual["validos"] /
    ndvi_mensual["registros"] * 100
)

print(
    ndvi_mensual.to_string(
        index=False,
        formatters={
            "porcentaje_valido": lambda x: f"{x:.1f}%"
        }
    )
)

print("\nAuditoría terminada; no se modificaron los datos originales.")
```

```python

# CELDA 27 — MATRIZ DE DISPONIBILIDAD POR SENSOR

from pathlib import Path
import pandas as pd

archivo = (
    Path(ROOT) / "data" / "raw" / "tabular"
    / "Conjunto_datos_BASICO_AgroCebada2026.csv"
)

df = pd.read_csv(archivo)

excluir = {
    "ID_POLIGONO",
    "fecha_captura",
    "sensor",
}

variables = [
    c for c in df.columns
    if c not in excluir
    and pd.api.types.is_numeric_dtype(df[c])
]

resumen = []

for sensor, grupo in df.groupby("sensor"):
    for variable in variables:
        resumen.append({
            "sensor": sensor,
            "variable": variable,
            "registros": len(grupo),
            "validos": int(grupo[variable].notna().sum()),
            "porcentaje_valido": (
                grupo[variable].notna().mean() * 100
            ),
            "valores_distintos": grupo[variable].nunique(
                dropna=True
            ),
        })

disponibilidad = pd.DataFrame(resumen)

tabla = disponibilidad.pivot(
    index="variable",
    columns="sensor",
    values="porcentaje_valido"
).fillna(0)

print("=== PORCENTAJE DE VALORES VÁLIDOS POR SENSOR ===")
print(tabla.round(1).to_string())

print("\n=== VARIABLES CON DATOS EXCLUSIVOS DE CADA SENSOR ===")

for sensor in sorted(df["sensor"].dropna().unique()):
    exclusivas = disponibilidad[
        (disponibilidad["sensor"] == sensor)
        & (disponibilidad["porcentaje_valido"] > 0)
    ]["variable"].tolist()

    print(f"\n{sensor}:")
    print(exclusivas)

print("\n=== VARIABLES COMPLETAMENTE VACÍAS ===")
for sensor in sorted(df["sensor"].dropna().unique()):
    vacias = disponibilidad[
        (disponibilidad["sensor"] == sensor)
        & (disponibilidad["validos"] == 0)
    ]["variable"].tolist()

    print(f"\n{sensor}: {vacias}")

print("\nAuditoría terminada; no se modificaron los datos originales.")
```

```python

# CELDA 28 — INVENTARIO DETALLADO DE RÁSTERES ORIGINALES

from pathlib import Path
from collections import defaultdict
import rasterio

raw_dir = Path(ROOT) / "data" / "raw"

rasteres = sorted(
    p for p in raw_dir.rglob("*")
    if p.is_file() and p.suffix.lower() in {".tif", ".tiff"}
)

grupos = defaultdict(list)

for p in rasteres:
    nombre = p.name.upper()

    if nombre.startswith(("PREC_", "PRECIP_")):
        grupo = "PRECIPITACION"
    elif nombre.startswith("TMAX_"):
        grupo = "TEMPERATURA_MAXIMA"
    elif nombre.startswith("TMIN_"):
        grupo = "TEMPERATURA_MINIMA"
    else:
        grupo = "OTROS"

    grupos[grupo].append(p)

print("=== CANTIDAD DE RÁSTERES POR GRUPO ===")
for grupo, archivos in sorted(grupos.items()):
    print(f"{grupo}: {len(archivos)}")

print("\n=== MUESTRA DE NOMBRES DE ARCHIVO ===")
for grupo, archivos in sorted(grupos.items()):
    print(f"\n{grupo}")
    for p in archivos[:10]:
        print(" ", p.relative_to(raw_dir))

print("\n=== METADATOS DE LOS PRIMEROS RÁSTERES ===")
for grupo, archivos in sorted(grupos.items()):
    if not archivos:
        continue

    p = archivos[0]

    with rasterio.open(p) as src:
        print(f"\nArchivo: {p.relative_to(raw_dir)}")
        print("CRS:", src.crs)
        print("Dimensiones:", src.width, "x", src.height)
        print("Bandas:", src.count)
        print("Tipo de datos:", src.dtypes)
        print("NoData:", src.nodata)
        print("Resolución:", src.res)
        print("Extensión:", src.bounds)

print("\nInventario terminado; no se modificaron los archivos.")
```

```python

# CELDA 29 — AUDITORÍA DE LAS VARIABLES DEL MODELO RF124

from pathlib import Path
import json
import joblib

model_path = (
    Path(ROOT) / "models" / "reconstruccion_rf124_leaf16"
    / "RF124_leaf16_final.joblib"
)

metadata_path = (
    Path(ROOT) / "models" / "reconstruccion_rf124_leaf16"
    / "metadata.json"
)

if not model_path.exists():
    raise FileNotFoundError(f"No se encontró el modelo: {model_path}")

modelo = joblib.load(model_path)

print("=== MODELO ===")
print("Tipo:", type(modelo).__name__)

# Recuperar nombres de variables, si el estimador los conserva.
feature_names = getattr(modelo, "feature_names_in_", None)

if feature_names is None and hasattr(modelo, "named_steps"):
    for nombre, paso in reversed(list(modelo.named_steps.items())):
        feature_names = getattr(paso, "feature_names_in_", None)
        if feature_names is not None:
            print("Variables recuperadas del paso:", nombre)
            break

if feature_names is None:
    print(
        "\nEl objeto no conserva feature_names_in_. "
        "Revisaremos sus metadatos o el notebook de reconstrucción."
    )
else:
    feature_names = list(feature_names)
    print("Número de variables:", len(feature_names))

    print("\n=== LISTA COMPLETA DE VARIABLES ===")
    for i, nombre in enumerate(feature_names, start=1):
        print(f"{i:03d}. {nombre}")

    # Clasificación preliminar por nombre; no sustituye una revisión.
    print("\n=== CLASIFICACIÓN PRELIMINAR ===")

    grupos = {
        "CLIMÁTICAS": [],
        "TOPOGRÁFICAS": [],
        "ESPECTRALES / TEMPORALES": [],
        "OTRAS / POR CLASIFICAR": [],
    }

    for nombre in feature_names:
        n = nombre.lower()

        if any(x in n for x in [
            "prec", "precip", "tmax", "tmin", "temperatura",
            "rain", "climate", "clima"
        ]):
            grupos["CLIMÁTICAS"].append(nombre)
        elif any(x in n for x in [
            "elev", "altitud", "pendiente", "slope", "dem", "topo"
        ]):
            grupos["TOPOGRÁFICAS"].append(nombre)
        elif any(x in n for x in [
            "ndvi", "evi", "lai", "savi", "msavi", "vi6t",
            "ndwi", "ndti", "fapar", "dswi", "mcrc", "nddi",
            "srndi", "landsat", "sentinel", "planet"
        ]):
            grupos["ESPECTRALES / TEMPORALES"].append(nombre)
        else:
            grupos["OTRAS / POR CLASIFICAR"].append(nombre)

    for grupo, nombres in grupos.items():
        print(f"\n{grupo}: {len(nombres)}")
        for nombre in nombres:
            print("  ", nombre)

if metadata_path.exists():
    print("\n=== METADATOS DEL MODELO ===")
    with open(metadata_path, "r", encoding="utf-8") as f:
        metadata = json.load(f)
    print(json.dumps(metadata, ensure_ascii=False, indent=2))
else:
    print("\nNo existe metadata.json.")
```
